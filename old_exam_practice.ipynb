```python
%%stata
use "Exam\PartD.dta" ,clear
```

## Step 1 - Exclude missing data

```python
%%stata
* No missing values so we can proceed to creating the dummy variables
drop if knowhow ==.
drop if comskill ==.
drop if perform ==.
drop if years ==. 
```

## Step 2 `a1` `a2` and `a3` calculation using the sequential way or `sureg`. I would highly advice for `sureg` to come into effect by the time when all its inputs are thoroughly known. So the manual calculation of `b` and `c'` is also required

```python
%%stata
* a(s) calculation
reg knowhow years
reg comskill years


* c' and b(s) calculations
reg perform years knowhow comskill 
```

```python
%%stata
* bootstraping
set seed 1234
capture program drop med

program multimed, rclass
sureg (knowhow years) (comskill years) (perform years knowhow comskill)
return scalar indknowhow = [knowhow]_b[years]*[perform]_b[knowhow]
return scalar indcomskill = [comskill]_b[years]*[perform]_b[comskill]
return scalar indtotal = [knowhow]_b[years]*[perform]_b[knowhow] + [comskill]_b[years]*[perform]_b[comskill]
end

bootstrap r(indknowhow) r(indcomskill) r(indtotal), reps(3000): multimed 
estat bootstrap, p bc
```

```python
%%stata
* spss will be picked as a reference category
codebook package
```

## Step 1 - Exclude missing data

```python
%%stata
* No missing values so we can proceed to creating the dummy variables
drop if package ==.
drop if easeofus ==.
drop if function ==.
drop if attitud ==. 
```

## Step 2 `a1` `a2` and `a3` calculation using the sequential way or `sureg`. I would highly advice for `sureg` to come into effect by the time when all its inputs are thoroughly known. So the manual calculation of `b` and `c'` is also required

```python
%%stata
* a(s) calculation
reg easeofus package
reg function package


* c, c' and b(s) calculations
reg attitud package easeofus function
reg attitud package
```

```python
%%stata
* bootstraping
set seed 1234
capture program drop multimed2

program multimed2, rclass
sureg (easeofus package) (function package) (attitud package easeofus function )
return scalar indease = [easeofus]_b[package]*[attitud]_b[easeofus]
return scalar indfunction = [function]_b[package]*[attitud]_b[function]
return scalar indtotal = [easeofus]_b[package]*[attitud]_b[easeofus] + [function]_b[package]*[attitud]_b[function]
end

bootstrap r(indease) r(indfunction) r(indtotal), reps(3000): multimed2
estat bootstrap, p bc
```
